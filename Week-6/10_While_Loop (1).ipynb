# What Is While Loop

* while statement repeatedly executes a block of statements as long as a test at the top keeps evaluating to a true  value \
  it is called a “loop” because control keeps looping back to the start of the statement until the test becomes false \
  when the test become false, control passes to the statement that follows the while block \
  if the test if false to begin with, the body never runs and the while statement is skipped

# Use while Structure

* <code>
  while condition:
    statement 
  </code>

##### keep printing hello

```python
# while True:
#     print("hello")
```

* because the condition is always True, Python keeps executing the body forever \
  it is called infinite loop \
  we can forcibly terminate it

##### keep slicing the first character

```python
message = "hello"

while message:
    # * message is True if it is not empty
    print(message)
    message = message[1:]
    
```

##### keep counting

```python
num = 1

while num <= 10:
    print(num)
    num += 1
```

```python
flag = False

while flag == True:
    print("Hello Wolrd !")
```

# Alter The Flow Of A while loop

## Use break Statement

* jumps out of the closest enclosing loop \
  past the entire loop statement \
  it is only used inside loops 
* <code> 
  while condition: 
    statements 
    if condition:
        break 
    statements
</code>
  break can appear anywhere inside the while loop but they are usually in if statement to take action in response to some condition

```python
target = 10
times = 0

while times < 3:
    guess_num = input("please guess a number")
    
    if int(guess_num) == target:
        print("Congratulations !")
        break
    
    times += 1
```

## Use continue Statement

* jumps to the top of the closest enclosing loop \
  jumps to the loop’s header line \
  it is only used inside loops
* <code> 
  while condition: 
    statements 
    if condition:
        continue 
    statements
</code>
  break can appear anywhere inside the while loop but they are usually in if statement to take action in response to some condition

```python
target = 10
times = 0

while times < 3:
    guess_num = input("please guess a number")
    
    if int(guess_num) == target:
        print("Congratulations !")
        break
    
    elif abs(int(guess_num) - target) > 10:
        continue
    
    times += 1
```

# while else Structure

* <code>
  while condition:
      statement
      if condition:
          break
  else:
      statement
  </code>
* the else block runs if and only if the loop is exited without hitting a break
* this serves a way to catch the search-failure case

##### determine whether a positive integer x is prime

```python
x = 36
potential_factor = x//2

while potential_factor > 1:
    if x % potential_factor == 0:
        print(f"{x} has factor {potential_factor}")
        print("it is not a prime")
        break
    
    potential_factor -= 1
    
else:
    print("do no found any facotr")
    print(f"{x} is a prime")
    
```

```python
x = 1
potential_factor = x//2

while potential_factor > 1:
    if x % potential_factor == 0:
        print(f"{x} has factor {potential_factor}")
        print("it is not a prime")
        break
    
    potential_factor -= 1
    
else:
    print("do no found any facotr")
    print(f"{x} is a prime")
```

* note that if the condition is False at the begining the body is never executed \
  so we will not run the break either \
  in this case the else block is also run \
  so this code fails for 0 and 1, as well as negative numbers

##### an alternative way: using flag

```python
x = 7
potential_factor = x//2
found_flag = 0

while (potential_factor > 1) and (not found_flag):
    if x % potential_factor == 0:
        print(f"{x} has factor {potential_factor}")
        print("it is not a prime")
        found_flag = 1
        break
    potential_factor -= 1

    
if not found_flag:
    print("do no found any facotr")
    print(f"{x} is a prime")
```

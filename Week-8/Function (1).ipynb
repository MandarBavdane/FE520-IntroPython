# What Is The Function

* a function is a device that groups a set of statemetns so they can be run more than once in a program 
* maximize code reuse and minimize redundancy \
  if we need to perform a specific task multiple times throught the program \\\
  we don't need to type all the code again and again \
  we just need to package a block of code dedicated to handling that task and assign this package of code a name \
  everytime we want to run the code inside the function we call the functiion 
* function can allow us specify parameters that serve as function inputs and may differ each time the code id run
* if we need to modify the function \\\
  we can change the code once and the change will take place everywhere the function is called
* decompose procedure \
  it is easier to implement the smaller tasks in solation than it is to implement the entire process at once \
  for example, to make a pizza from scratch we would \
  l1: mixing a dough \
  l2: rolling it out \
  l3: adding toppins \
  l4: baking it \
  function lets us split complex systems into pieces that have well-defined roles \
  one function is for each subtask in the process
  

# Define A Function
<code> 
def name(parm1, parm2, ..., parmN):
    statements
</code>

* using the keyword def is to inform Python that we are defining a function \
  then we set a name of function followed by a parentheses \
  if the function needs no information to do its job, the parentheses is empty \
  if the function needs information, we pass information the functions needs to do its job \ \
  by adding the variable we allow function to accept any value of the variables we specify \
  the variable in the function definition is called parameter \
  then we put a colon followed by the body of the function \
  the colon creates a function block \
  any indented lines that follow the definition make up the body of the function
  
* def executes at runtime \
  when Python reaches and this def, the def statements creates a function object and assigns it to a name \
  your function does not exist until Python reaches and runs the def
  
<code>
name(arg1, arg2, ..., argN)
</code>

* when we want to use the function we call it it tells the Python to run the code inside the function \
  we type in the name followed by parentheses \
  inside the parentheses we pass the information that function needs to do its job \
  the value is called argument \
  if we did not add any variable in function definition, the parenthese is empty \
  if we add a variable in function definition \
  now the function expects us to provide a value for the variable \
  arguments are passed by assignment \
  the argument names in the header are assigned to the objects passes in the parentheses at the point of call \
  now we can call the function as often as we want and pass it any value

```python
def square(x):
    """square a number"""
    # * it is called docstring
    #   it describes what the function does
    #   this string is enclosed in triple quotes
    #   when Python generate the documentation for the function, Python looks for this

    result = x**2 
    
    print(result)
    # * Python can process some data and then return the value
```

```python
help(square)
```

```python
square(3)
```

# Pass Argument

## Use Positional Argument

```python
def power(x, y):
    result = x**y
    return result
```

```python
power(3,2)
# * if we use positional arguments \
#   it needs to be in the same order the parameters were written
#   Python must mathc each argument in the function call \
#   with the corresponding parameter in the function definition
```

```python
power(2,3)
```

* order matters in positional arguments \
  we can get unexpected results if we mix up the order of the arguments in the function call

## Use Keyword Argument

```python
def power(x, y):
    result = x**y
    return result
```

```python
power(y = 2, x = 3)
# * if we use keyword argument \
#   each keyword argument consists of a variable name and a value
# * when we call the function \
#   we explicitly tell Python which parameter each argument should be matched with
#   in this way there is no confusion and Python knows where each argument should go
```

```python
power(x = 3, y = 2)
```

* the order of keyword arguments doesn't matter

```python
power(3, y = 2)
#power(x=3, 2)
```

* positional argument must be before keyword argument

```python
#power(3)
#power(3,4,5)
```

* when we provide fewer or more arguments than a function needs to do its work \
  we will get an unmatched error

## Set Default Value

```python
def power(x, y = 2):
    # * we can define a default value for each parameter
    #   in the function call \
    #   if an argument is provided, Python uses the argument value
    #   if not, Python uses the default value
    # * when we use default values \ 
    #   any parameter with a default value needs to be listed \
    #   after all the parameters that don't have default value
    #   this allows Python to continue using positional argument correctly
    result = x**y
    return result
```

```python
power(3)
```

* now anyone who want to square the number can omit the 2

## Pass An Arbitrary Number of Arguments

```python
def register_language(*language):
# * we put an asterisk before a parameter
#   it collects as many arguments as the user provides
#   it tells Python to make an empty tuple called language \
#   and pack whatever values it receives into this tuple
# * we will often see the generic parameter name *args

    print(len(language))
    print(language)
```

```python
register_language("Python", "C++", "R")
```

```python
register_language("Python")
```

```python
def student_register_language(student, *language):
    
    print(student)
    print(len(language))
    print(language)
```

```python
student_register_language("student A", "C++", "Python")
```

* the parameter that accepts an arbitrary number of arguments must be placed last in the function definition \
  Python matches positional arguments and keyword arguments first \\\
  and then collects any remaining arguments in the final argument

## Pass Arbitrary Keyword Arguments

```python
def collect_information(**student_info):
# * we put double asterisks before parameter
#   it collects as many name-value pairs as the user provides
#   it tells Python to make an empty dictionary called student_info \
#   and and pack whatever name-value pairs it receives into this dictionary
# * we will often see the parameter name **kwargs

    print(student_info)
    print(student_info["cwid"])
```

```python
collect_information(cwid=100, major="FE", GPA=4, writing_award=True)
```

# Return Value

* function boies oftern contain a return statement \
  the return statement can show up anywhere in a function body \
  when reached it ends the function call and sends a result back to the caller \
* if the value is omitted, return sends back a None

## Return A List

```python
def square_seq(list_a):
    result = []
    for i in list_a:
        result.append(i**2)
    
    return result
```

```python
list_a = [1,2,3,4,5]
square_seq(list_a)
```

## Return A Dictionary

```python
def collect_student_info(student_name, score):
    student_info = {student_name: score}
    
    return student_info
```

# What Is The String

* string are used to represent textual information including: \
  (1) word \
  (2) symbol \
  (3) website address \
  &ensp;&ensp;&ensp;&ensp; e.g., https://www.google.com/ \
  (4) file path in computer \
  &ensp;&ensp;&ensp;&ensp; e.g. windows file path C:\Users\username\Documents
  
  
* string are sequence of characters \
  in other words, string are ordered collection of characters \
  sequence maintains a left-to-right positional order among items they contain \
  each item has its own relative position

### Unicode Strings

* it let us embed characters in strings that cannot easily be typed on a keyboard
* to store the human-readable characters on computers we need to encode them into 0/1 \
  we also need to decode the 0/1 into human-readable characters for representation \
  one 0/1 digits is called bit \
  combine 8 0/1 digits we get a byte \
  the rule that defines the encoding process is called encoding shcema \
  l1: ASCII \
  l2: unicode \
  l3: UTF-8 
* ASCII (American Standard Code for Information interchange) \
  it is a common character encoding format for text data in computers \
  there are unique valuess for 128 alphabetic, numeric, or special additional characters and control codes \
  it includes upper and lower case letters A through Z, numerals 0 through 9, basic punctuation symbols 
* representation \
  ASCII characters may be represented in the following ways: \
  l1: hexadecimal \
  l2: octal \
  l3: binary (7 bit or 8 bit) 0100 0001 \
  e.g. character A \
  hexadecimal: 0x41\
  octal: 101 \
  binary: 0100 0001 
* Unicode is used to deal with non-ASCII text \
  ASCII is a part of Unicode \
  it is required for processing text in internationalized character sets \
  it is especially useful in internet domain like web page, email content \
  e.g. characters in the Japanese alphabets are outside the ASCII set \
  https://www.techonthenet.com/unicode/chart.php


```python
print("\x41")
# * characters can be represented as backslash escape sequences 
# * use backslash followed by a hexadecimal digits
#   \xNN hexadecimal

print("\101")
# * use backslash followed by a octal digits  
```

```python
print("\x41")
print("\xc4")

print(u'\u0041')
print(u"\u00c4")
# * u"": unicode string literal
```

# Sequence Operation

* string is one type of sequence \
  we can access thier components by position

##  Indexing

```python
message = "Hello World !"

print(message[0])
# * message: we write the name of the string
# * []: we use square brackets to slice
#       inside of the square brackets we can refer to character by their index
# * 0: the index start at 0
#      python considers the first item to be at index 0
#      not index 1
```

```python
message = "Hello World !"

print(message[1])
```

```python
message = "Hello World !"

print(message[12])
print(message[-1])
# * by telling python the index -1, we can access the last element of a string
#   in this way, we can access the last element without knowing exactly how long the string is
```

```python
message = "Hello World !"

print(message[-2])
# * we can extend -1 to other negative index as well
```

## Slicing

* it allow us to extract a substring

```python
message = "Hello World !"

print(message[0: 5])
# * slice operator[start_index : end_index]
#   we can specify a range
#   slice operator will grab the element from start index up to end index \
#   but do not include the end index
#   that is, it will stop one item before the end index we specify
# * if the start index is omitted, it is 0 by default
#   if the end index is omitted, it is set to end of the string
#   for the backward index, if the end index is omitted, it is set to the beginning of the string
```

```python
message = "Hello World !"

print(message[ : ])
print(message[1: ])
```

```python
message = "Hello World !"

print(message[ : : 2])
# * we can specify the third parameter: step
#   it allows us for skipping items

print(message[ : : -1])
# * negative indexes means to count backward from the end or right
# * setting step to -1 allows us to revert the string 

print(message[0: 5: -1])
print(message[5: 0: -1])
# * with a negative step, the meanings of the start and the end are reversed as well
```

# String Method

* methods are actions or behaviors that are attached to and act on a specific object \
  they are triggred with a call expression
* method call syntax \
  object.method( ) \
  fetch the method in object \
  then it will pass object itself as the first information and zero or more information into parentheses as needed \
  it means we apply the action with information on this object itself

## Implement .lower( ) Method

```python
message = "Hello World !"

print(message)
print(message.lower())
# * message: we write the name of variable
# * .: the dot after the name of variable tells python to make the method act on itself
# * (): every method is followed by parentheses 
#       it indicates it is an action
#       we can put additional information into it
# * .lower(): change a string to all lowercase
```

## Implement .upper( ) Method

```python
message = "Hello World !"

print(message)
print(message.upper())
# * .upper(): change a string to all uppercase
```

## Implement .rstrip( ) Method

```python
message_1 = "Hello World !"
message_2 = "Hello World ! "

print(message_1)
print(message_2)
```

* whitespace refers to any nonprinting character \
  e.g. space, tab, end-of-line symbol
* extra whitespace may be confusing in program \
  to programmers, the output of "Hello World !" and "Hello World ! " look pretty much the same \
  to program, Python detects the extra space and consider them as two different strings

```python
message_1 = "Hello World !"
message_2 = "Hello World ! "

print(message_1)
print(message_2.rstrip())
```

## Implement .index( ) Method

```python
message = "Hello World !"

print(message.index("H"))
# * index(): find the index of a character or a sequence of characters in a string 
# * we pass a parameter
#   the character or sequence of character whose index we are looking for
```

```python
message = "Hello World !"

print(message.index("World"))
```

* it returns the index of the start of World in the string

```python
message = "Hello World !"

print(message.index("o"))
print(message.index("o", 5))
```

* it returns the index of first o in the string by default

## Implement .replace( ) Method

```python
message = "Hello World !"

print(message.replace("World", "Jack"))
# * .replace(): replace a character or a sequence of characters with another character or a sequence of characters
```

```python
message = "Hello World World World!"

print(message.replace("World", "Jack", 2))
# * the third parameter count: it specifies how many occurrences of the old value we want to replace 
#   default is all occurrences
```

```python
message = "%hello%world"

print(message.replace("%", ""))
```

## Implement .find( ) Method

```python
message = "Hello World !"

print(message.find("W"))
print(message.find("h"))

# * .find(): it is a substring search operation
#   it returns index of the substring or -1 if it is not present
```

## Implement .split( ) Method

# Immutability

* strings are immutable \
  We cannot change the value of immutable objects after we created.
* despite all of string methods, we are not changing the original string, but creating new strings as the results \
  such results will not be retained \
  this is if we print the variable we will get the original results \
  if we want to retain the new results, we can assign it to a variable \
  of course we can always assign it back to the original variable 

```python
message = "hello"
text = message
message = "aello"

print(message)
print(text)
```

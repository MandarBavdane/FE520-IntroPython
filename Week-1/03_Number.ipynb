# What Is The Number

* some numeric data types in python \
  (1) int \
  &emsp;&ensp; integer is a whole number \
  (2) float \
  &emsp;&ensp; float is a decimal number \
  &emsp;&ensp; python calls any number with a decimal point a float \
  (3) complex \
  &emsp;&ensp; a complex number is a number with real and imaginary components

# Literals

## Integer

```python
print(3)

print(type(3))
# * we can use the type() function to know which type a value belongs to
```

```python
print(1_0_0)
# * _: python uses undersocre to connect digit 
#      to python, 1_0_0 is the same as 100
#      when writing long number this will make the number more readable
```

## Float

```python
print(3.5)

print(type(3.5))
```

## Complex Number

```python
print(1 + 7j)
# * Python complex literals are written as real_part + imaginary_part
print(type(1 + 7j))

print(7j)
print(type(7j))
# * the imaginary_part may be appear on its own
#   the real_part is optional 
```

## Hexadecimal, Octal, And Binary Literals

* all of these liteals produce integer objects \
  they are just alternative syntaxs for specifying valus

```python
print(0x9)
# * hexadecimals start with a leading 0x or 0X, followed by hexadecimal digits

print(0O011)
# * octals start with a leading 0o or 0O, followed by octal digits

print(0b1001)
# * binary start with a leading 0b or 0B, followed by binary digits.
```

```python
print(hex(9))
print(oct(9))
print(bin(9))
print(int("0x9", 16))
# * int(str, base): convert a string to an integer per a given base
```

# Define A Variable

```python
x = 1

print(x)
```

##### Take Multiple Assignment

```python
x, y, z = 1, 2, 3
# * we can assign value to more than one variable using just a single line
#   separate the variable names with commas \
#   besides seprate the value with commas \
#   the number of variable matches the number of values
#   python will assgin value to its respectively positioned variable


print(x)
print(y)
print(z)
```

# Numeric Operation

##### Find The Quotient In Divident

```python
print(10//3)
```

##### Find The Remainder In Divident

```python
print(10%3)
# * %: it is modulus operator
```

##### Consider Operator Precedence

```python
print(1*2 + 3*4)
# * code complex expressions by stringing together the operator expressions
#   operator precedence determines which operation to perform first

print(1*(2+3)*4)
# * parentheses group subexpressions
#   use parentheses to override precedence rules
#   Python always evaluates expressions in parentheses first
#   force Python to evaluate the expression in the desired order
```

# Number Comparison

```python
print(1<2<3)
# * comparison operators may be chained x<y<z
#   it produces the same results as x<y and y<z
```

```python
print(1.1 + 2.2 == 3.3)
print(1.1 + 2.2)
```

# Use Increment

```python
x = 1

x = x + 1

print(x)
```

```python
x = 1

x += 1
# * +=: it is an augmented assignment operator
#       it is equivalent to x = x + 1

print(x)
```

# Deal With Number With Function

## Implement round( ) Function

```python
x = 3.5

print(round(x))
```

## Implement abs( ) Function

```python
x = -3.5

print(abs(x))
# * abs(): it will return an absolute value
```

## Implement pow( ) Function

```python
x = 2
print(pow(2, 3))
# * pow(): take power
# * the first parameter is the number
# * the second parameter is the power
```

## Implement max( ) Function

```python
print(max(2, 3))
# * it returns the maximum value
```

## Implement min( ) Function

```python
print(min(2,3))
# * it returns the minumum value
```

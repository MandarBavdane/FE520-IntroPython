# What Is The Dictionary

* dictionary are used to represent collections \
  the collection's item are named to describe properties \
  l1: student information with its property \
  l2: search table
* dictionaries are not sequences \
  they are unordered collections \
  they do not matintain any dependable left-to-right order \
  sequence operations do not work
* a dictionary in Python is a mapping of key-value pairs \
  each value is associated to a key \
  we can grow and shrink on demand \
  the value can be a number, a string, a list, or even another dictionary

# Dictionary Literal

## Plain Dictionary

```python
print({"student A": 98, "student B": 89, "student C": 73, "student D": 37})
# * in Python curly brackets indicate a dictionary
#   inside the curly brackets we input a series of key-value paris
#   we use a colon to connect key and value
#   each key-value pair is separated by comma
```

* even if it shows some order, it is not fixed \
  the order of keys can be different from wat we originally typed \
  key order is arbitrary \
  it might vary per release, platform, and interactive session \
  so we should not depend on the order of the key

```python
student_grade = {"student A": 98, "student B": 89, "student C": 73, "student D": 37}
print(student_grade)
```

```python
student_grade = {1: 98, 2: 89, 3: 73, 4: 37}
# * so far we have used strings as keys, but keys need not always be strings \
#   any immutable objects can be keys

print(student_grade)


movie_database = {(2000, "alice") : "movie 1", 
                  (2001, "bob") : "movie 2",
                  (2002, "cathry") : "movie 3"}
# * tuples can be used as dictionary keys too
#   it allows to compound keys
print(movie_database)
```

```python
student_grade = dict([("student A", 99), 
                      ("student B", 88)])
# * this way is useful if we need to build up keys and values as sequences at runtime
print(student_grade)
```

```python
student_grade_zip = zip(["student A", "student B"], [99, 99])
# * zip() functions allow to combind separate lists of keys and values obtained dynamically at runtime
print(list(student_grade_zip))

student_grade_zip = zip(["student A", "student B"], [99, 99])
student_grade = dict(student_grade_zip)
print(student_grade)
```

# Access Element In A Dictionary

## Use Key

```python
student_grade = {"student A": 98, "student B": 89, "student C": 73, "student D": 37}

print(student_grade["student A"])
# * because dictionary is collection of key-value pairs, we can fetch the item by keys
#   we still use slice operator
#   but the item in the square brackets is a key, not a relative position
```

## Modify Element In A Dictionary

### Modify Value In A Dictionary

```python
student_grade = {"student A": 98, "student B": 89, "student C": 73, "student D": 37}

student_grade["student A"] = 100

print(student_grade)
```

### Add New Key-value Pairs

```python
student_grade = {"student A": 98, "student B": 89, "student C": 73, "student D": 37}

student_grade["student E"] = 50
# * whenever we assign a new dictionary key (one that has not been assigned before) \
#   we create a new entry in the dictionary.

print(student_grade)
```

### Remove Key-value Pairs

```python
student_grade = {"student A": 98, "student B": 89, "student C": 73, "student D": 37}

del student_grade["student D"]

print(student_grade)
```

# Deal With Dictionary With Method

## .get( ) Method 

```python
student_grade = {"student A": 98, "student B": 89, "student C": 73, "student D": 37}

print(student_grade.get("student A", "the key doesn't exist"))
# * .get(): retrieve value
# * parameter 1: key
# * parameter 2: pass the value to be returned if the key doesn't exist
```

* in this case the key student A exists, so we get the corresponding value

```python
student_grade = {"student A": 98, "student B": 89, "student C": 73, "student D": 37}

print(student_grade.get("student Z", "the key doesn't exist"))
```

* in this case the key student Z doesn't exists, so we get the hint instead of an error

## .keys( ), .values( ), .items( ) Methods

```python
student_grade = {"student A": 98, "student B": 89, "student C": 73, "student D": 37}

print(student_grade.keys())

#print(student_grade.keys()[1])
# * this will return a bag of values called view object, instead of a physical list
#   it is not subscriptable, which means we cannot access it using index

print(list(student_grade.keys())[1])
# * the list() function forces it to produce all its value at once


print(student_grade.items())
# * it returns a bag of key-value pair tuples
print(student_grade.values())
```

```python
student_grade = {"student A": 98, "student B": 89, "student C": 73, "student D": 37}
print(student_grade)

key = student_grade.keys()
print(list(key))

del student_grade["student C"]
print(student_grade)
print(list(key))
```

* the change in the dictionary will be reflected in any current view objects

```python
student_grade = {"student A": 98, "student B": 89, "student C": 73, "student D": 37}
print(student_grade)

key_list = list(student_grade.keys())
print(list(key_list))

del student_grade["student C"]
print(student_grade)
print(key_list)
```

## .update( ) Method

```python
student_grade = {"student A": 98, "student B": 89, "student C": 73, "student D": 37}
print(student_grade)

add = {"student E": 77}
student_grade.update(add)
# * it merges the key-value pairs of one dictionary into another
print(student_grade)

add_more = {"student F": 88, "student G": 99}
student_grade.update(add_more)
print(student_grade)


add = {"student D": 100}
student_grade.update(add)
# * if we attempt to update a key that existed already, the original one will be overwritten
print(student_grade)
```

## .pop( ) Method

```python
student_grade = {"student A": 98, "student B": 89, "student C": 73, "student D": 37}
print(student_grade)

get = student_grade.pop("student B")
print(student_grade)
print(get)
```

# Merge Two Dictionaries

```python
student_grade_1 = {"student A": 98, "student B": 89, "student C": 73}
student_grade_2 = {"student X": 10, "student Y": 20, "student Z": 30}

student_grade_total = {**student_grade_1, **student_grade_2}
# * ** is for unpacking dictionary

print(student_grade_total)
```

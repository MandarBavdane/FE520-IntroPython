# How To Name Variable

* variable names can contain only letters, numbers, underscores \
  they can start with a letter or an underscore \
  but not with a number \
  e.g. name , _ \
  &emsp;&emsp; name_ , name1 , \_name , _1 \
  &emsp;&emsp; name_1 , name1_\ , _name1 , _1name \
  &emsp;&emsp; but not 1 \
  &emsp;&emsp;&emsp;&emsp;&emsp;&emsp; 1name , 1_\
  &emsp;&emsp;&emsp;&emsp;&emsp;&emsp; 1_name , 1name_\
  &emsp;&emsp; underscores can be used to separate words in variable name \
  &emsp;&emsp; spaces are not allowed in variable name
* avoid using python keywords and function names as variable name \
  python has reserverd some words for a particular programmatic purpose \
  e.g. if

# What Is If Statement

* If statement allow the program to respond one way if certain conditions are true \\\
  and responds in a different way if those conditions are not true

# Use if Sturcture

## Structure

* <code>
  if conditional_test:
    statement
  </code>

* if the conditional test evaluates to True, Python executes the code following the if statement \
  if the conditional test evaluates to False, Python ignores the code following the if statement

```python
score = 90

if score >= 60:
# * when we use if statement we need to give computer a couple pieces of information
#   L1: use keyword if
#       when python sees the keyword, it knows we are going to make a decision
#   L2: conditional test
#       conditional test can be evaluated as True or False
#       Python uses the values True and False \
#       to decide whether the code in an if statement should be executed
#       if a conditional test evaluates to True \
#       Python executes the code following the if statement
#       if a conditional test evaluates to False \
#       Python ignores the code following the if statement
#   L3: add colon
#       colon is to tell python the next line is the start of codes that belong to the loop
#       it used to represent an indented block

    print("pass")
    # * we want to write a few lines of codes that belong to the if statement
    # * in order to tell Python that the statement belongs to a particular block of code 
    #   it is necessary and mandatory to use indentation
    #   indentation is the leading whitespace before any statement 
    #   python treats the statements that have the same indentation level \
    #   (i.e. same number of leading whitespace) \
    #   as a single block of code
    # * people often use tabs to indent
    # * python uses 4 spaces as default indentation spaces
    #   but the number of spaces can be anything and it is up to the user
    #   but a minimum space number is 1 
    #   besides, the number of spaces must be uniform in a block of code
    # * all the statements indented form a block and belong to the if statement
    #   if we write some codes without indentation, \
    #   python will be no longer considering them inside of the if statement
    #  Python does not care how you indent the code, but it cares that it is done consistently
```

```python
score = 90
print(score >= 60)
```

```python
score = 90
print(score == 90)
# * ==: it is equality operator
#       it checks whether two values are equal
#       if the values on the left side is equal to the right side \
#       it returns true
#       if the values on the left side is not equal to the right side \
#       it returns false
```

```python
score = 90
print(score != 90)
# * !=: it is inequality operator
#       it checks whether two values are not equal
```

```python
enrolled = True
performance = "good"

if enrolled:
    print("the student status is enrolled")
    
    if performance == "good":
        print("the student performance is good")
    
```

* this code contains three blocks: \
  l1: the top-level code of file is not indented at all \
  l2: the one within the outer if statement is indented four spaces \
  l3: the onw within the nested if is indented eight spaces

## Conditional Test

### Conditional Test Is Able To Be Reducible To Boolean Value

```python
if "no":
    print("yes")
    
if []:
    print("empty list")
```

* all objects have an inherent Boolean True or False value
* any nonzero number or nonempty object is True \
  Python will consider the following empty object as False \
  l1: empty string \
  l2: empty list \
  l3: None \

```python
print(bool("no"))
# * bool() function returns a True or False value as though it were tested in an if
print(bool([]))
```

### Use Logic Operator In Condition

* logic operators in Python are and, or, not
* the and and or operators always return an object - either the object on the left side of the operator or the object on the right \
  we will not get back a simple True or False \
  and and or operations evaluates the operand objects from left to right and stop as soon as the reuslt is known \
  this is usually called short-circuit evaluation \
  l1: or will stop and returns the first one that is True \
  l2: and will stop and returns the first one that is False
  

```python
a = True
b = False

print(a and b)
# * and: it checks whether two conditions are both True simultaneously
#        if both tests passes, the overall expression evaluates to True
#        if either test fails or if both tests fail, the overall conditional test evaluates to False
print(a or b)
# * or: it checks whether one of conditions are True
#       if either test passes or if both tests passes, the overall conditional test evaluates to True
#       if both tests fail, the overall conditional test evaluates to False
print(not a)
```

```python
print(2 or 3, 3 or 2)
print([] or 3)
print([] or {})
```

```python
print([] and 3, [] and {})
print(2 and 3, 3 and 2)
```

```python
score = 90
has_experience = True

if score >= 90 and has_experience:
    print("hire")
else:
    print("thank you")
```

# Use if-else Structure

## Structure

* if conditional test: \
  &emsp; statement \
  else: \
  &emsp; statement
* if the conditional test evaluates to True, Python will execute the code following the if statement \
  if the conditional test evaluates to False, Python will execute the code following the else statement 
* else statement is a catchall statement

```python
score = 59

if score >= 60:
    print("pass")
else:
    print("fail")
```

## if-else Ternary Expression

```python
gender = 0

if gender:
    baby_name = "jack"
else:
    baby_name = "jathy"

print(baby_name)
```

```python
gender = 1

baby_name = "jack" if gender else "jathy"
# * python runs expression "jack" only if gender turns out to be True, otherwise runs expression "jathy"

print(baby_name)
```

# Use if-elif-else Structure

* if conditional test 1: \
  &emsp; statement \
  elif conditional test 2: \
  &emsp; statement \
  else: \
  &emsp; statement
* if the conditional test 1 evaluates to True, Python will execute the code following the if statement \
  and Python will skip the rest of the conditional test \
  if and only if the conditional test 1 evaluates to False \\\
  Python will execute the code following the elif statement \
  if both the contitional test 1 and conditional test 2 evaluate to False \
  Python will execute the code following the else statement

```python
score = 90

if score >= 90:
    print("A")
elif 60 <= score < 90:
    print("B")
else:
    print("F")
```

* note that Python executes only one block in different kind of if structure \
  it runs each conditional test in order until one passes \
  when a test passes, Python will execute the code following that condition \\\
  and Python will skip the rest of the condition 

```python
score = 90

if score >= 90:
    print("A")
elif 80 <= score < 90:
    print("B")
elif 70 <= score < 80:
    print("C")
elif 60 <= score < 70:
    print("D")
else:
    print("F")
```

* we can use as many elif blocks as we like

```python
score = 90

if score >= 90:
    print("A")
elif 80 <= score < 90:
    print("B")
elif 70 <= score < 80:
    print("C")
elif 60 <= score < 70:
    print("D")
elif score < 60:
    print("F")
```

* sometime using elif statement is already clear enough \
  it can catch the specific condition \
  in this case we omit the else block at the end of an if-elif chain
* else statement is a catchall statement \
  it match any condition that wasn’t matched by a specific if or elif test \\\
  and that can sometime include invalid cases
  use the elif block and omit the else block will gain extra confidence \\\
  that your code will run only under the correct conditions

# Multiway Branching

## Use Multiple if statement

```python
score_A = 90
score_B = 89
score_C = 60

if score_A >= 90:
    print("student A got wonderful grade")

if score_B >= 90:
    print("student B got wonderful grade")
    
if score_C >= 90:
    print("student C got wonderful grade")
```

```python
choise = "spam"

if choise == "spam":
    print("1.25")
if choise == "ham":
    print("1.35")
if choise == "egg":
    print("0.99")
if choise == "bacon":
    print("1.10")
else:
    print("do not offer")
```

## Dictionary-Based Multiway Branching 

```python
choise = "spam"

food_price_dict = {"spam": 1.25, 
                   "ham": 1.35, 
                   "egg": 0.99, 
                   "bacon": 1.10}

print(food_price_dict[choise])

print(food_price_dict.get(choise, "do not offer"))
```

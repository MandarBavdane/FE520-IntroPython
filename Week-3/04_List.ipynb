# What Is The List

* lists are used to represent arbitrary collections \
  l1: lists of files in a folder \
  l2: lists of employees in a company \
  l2: lists of emails in a box
* lists are sequence of arbitrarily typed object \
  in other words, lists are ordered collection of arbitrarily typed object \
  sequence maintains a left-to-right positional order among items they contain \
  each item has its own relative position \
  Because strings are sequences, lists support all the sequence operations we discussed for strings \
  l1: lists have no fixed type constraint \
  &ensp;&ensp;&emsp;lists can contain elements of completely different data types \
  l2: lists have no fixed size \
  &ensp;&emsp; we can grow and shrink on demand

## Use list( )

```python
print(list(range(1,11)))
# * range(): create a sequence of numbers 
# * range(start_num, stop_num, stride)
#   the sequence of numbers starts at start_num \
#   and increments by stride repeatedly \
#   and stops at stop_num \
#   but do not include the stop_num
#   if the start_num is omitted, it is 0 by default
#   if the stride is omitted, it is 1 by default

# * list(): this function returns a list
# * signle parameter: we can pass a sequence into the function
#                     if no parameter is passed, it returns an empty list
```

```python
print(list(range(10)))

print(list(range(1,11,2)))
```

## Use Slicing

### Do Not Consider Stride

```python
goal = ["excellent", "good", "okay", "mediocre", "don't care"]

print(goal[-3:])
```

* the output retains the structure of the list

### Consider Stride

```python
goal = ["excellent", "good", "okay", "mediocre", "don't care"]

print(goal[0:6:2])
# * slice operator[start_index, end_index, stride]
#   stride parameter specifys the step of the slicing
#   if the stride is omitted, it is 1 by default
```

```python
numbers = list(range(10))
print(numbers)


print(numbers[::-1])
# * negative number for stride: it means the slicing is from right to left in inverse order 
```

* we can use slicing to get a reverse list

## Modify Element In A List

```python
goal = ["excellent", "good", "okay", "mediocre", "don't care"]
print(goal)

goal[0] = "wonderful"
print(goal)
```

* the output shows that the first item has indeed been changed \
  the rest of the list stay the same \
  it overwrites its former value

```python
goal = ["excellent", "good", "okay", "mediocre", "don't care"]
print(goal)
goal[0:2] = ["wonderful", "quite good"]
# * we can change list contents by assigning to an entire slice
print(goal)


goal = ["excellent", "good", "okay", "mediocre", "don't care"]
print(goal)
goal[0:2] = ["wonderful"]
# * the number of item inserted does not have to match the number of items deleted
# * we can think it as a combination of two steps:
#   l1: deletion
#       delelte the spcify to the left of the =
#   l2: insertion
#       the items contained in the collection to the right of =
print(goal)
```

```python
goal = ["excellent", "good", "okay", "mediocre", "don't care"]
print(goal)
goal[0:2] = []
# * by assigning the first two elements with an empty value, it is like deleting
print(goal)


goal = ["excellent", "good", "okay", "mediocre", "don't care"]
print(goal)
goal[0] = []
# * it is not deleting like the slice
#   it is assigning an empty list to the first index
print(goal)
```

# Deal With List With Method

## Implement .append( ) Method

```python
goal = ["excellent", "good", "okay", "mediocre", "don't care"]
print(goal)

goal.append("bad")
# * .append(): add new element to the end of the list
# * the first element: specify the value we want to add
print(goal)
```

```python
goal = ["excellent", "good", "okay", "mediocre", "don't care"]
print(goal)

goal = goal.append("bad")
# * .append() method will change the original list directly rather than creating a new list
#   so reassign is unnecesssary
# * moreover, it is not only unnecessary, it is incorrect
#   if we try to reassign back, we will get none
#   the .append() method return None value
print(goal)
```

## Implement .insert( ) Method

```python
goal = ["excellent", "good", "okay", "mediocre", "don't care"]
print(goal)

goal.insert(1, "pretty good")
# * .insert(): add a new element to any position of the list
# * the first parameter: specify the index we want to add
# * the second parameter: specify the value we want to add
print(goal)
```

* Python opens a space at index 1 and stores the value "pretty good" at that location \
  then shift every other value one position to the right

## Implement .pop Method

```python
goal = ["excellent", "good", "okay", "mediocre", "don't care"]
print(goal)

goal.pop()
# * .pop(): it removes the last item in a list
#           but we still can use the value after removing it
# * inside the parentheses we can specify the any index whose element we want to remove and grab
#   if we do not specify any index it will pop the last item in a list
print(goal)
```

```python
goal = ["excellent", "good", "okay", "mediocre", "don't care"]
print(goal)

pop_element = goal.pop()
           
print(pop_element)
```

* we have access to the value that was removed \
  the "don't care" is now assigned to the variable pop_element

```python
goal = ["excellent", "good", "okay", "mediocre", "don't care"]
print(goal)

goal.pop(1)
# * the .pop() method accepts an optinal offset of the item to be deleted and returned
print(goal)
```

## Implement .remove( ) Method

```python
goal = ["excellent", "good", "okay", "mediocre", "don't care"]
print(goal)

goal.remove("don't care")
# * .remove(): it removes an item in a list
# * inside the parentheses we pass by value of the item
print(goal)
```

```python
goal = ["good", "okay", "good", "good"]
print(goal)

goal.remove("good")
print(goal)
```

* the remove method deletes only the first occurrence of the value \
  If there is a possibility the value appears more than once we will need to remove it several times

## Implement .sort( ) Method

```python
score = [37, 100, 89, 98, 77]
print(score)

score.sort()
# * .sort(): sort the list 
# * parameter 1: reverse 
#                sort method will sort the list in ascending order by default
#                when set reverse = True it sort the list in descending order
# * parameter 2: key
#                we can define a function as sorting criteria
print(score)
```

```python
score = [37, 100, 89, 98, 77]
print(score)
score.sort(reverse = True)
print(score)


goal = ["excellent", "good", "okay", "mediocre", "don't care"]
print(goal)
goal.sort()
print(goal)
```

* the list of goal now is in alphabetical order

```python
letter = ['g', "A", "a", "E", "G", "e"]
print(letter)
letter.sort()
print(letter)

letter = ['g', "A", "a", "E", "G", "e"]
print(letter)
letter.sort(key = str.lower)
# * the key arguments provide an instruction what value should be used in sorting
#   we specify the name of the method we want to use to transform the value during the sort
#   we do not need the parenthesis
print(letter)


score = [-1, 2, -3]
print(score)
score.sort()
print(score)

score = [-1, 2, -3]
print(score)
score.sort(key = int.__abs__)
print(score)
```

```python
goal_score_comparison = [["excellent", 100], ["good", 98], ["okay", 89], ["mediocre", 73], ["don't care", 37]]
print(goal_score_comparison)

goal_score_comparison.sort()
print(goal_score_comparison)
```

## Implement .reverse( ) Method

```python
score = [37, 100, 89, 98, 77]
print(score)

score.reverse()
# * .reverse(): reverse the order of list
print(score)
```

* notice that .reverse() method doesn't sort the value \
  it simply reverses the order of list
